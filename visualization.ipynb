```python
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
TASKS = ['robin', 'killer']

level_values = {}

for task in TASKS:
    levels_dir = Path(f'./levels/Assignment03LevelGenerator/{task}')
    for file_path in levels_dir.rglob('*'):
        if file_path.is_file() and file_path.name.endswith('-values.txt'):
            level_name = f'{task}-{file_path.name.split('-values.txt')[0].split('lvl-')[-1]}'
            with file_path.open('r') as file:
                level_values[level_name] = list(map(float, file.readline().strip().split()))
```

```python
for level_name in sorted(level_values.keys()):
    plt.plot(level_values[level_name])
    plt.title(level_name)
    plt.xscale('log')
    plt.show()
```
